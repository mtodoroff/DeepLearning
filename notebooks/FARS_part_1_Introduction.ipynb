# Financial Asset Recommendation System


## 1. Problem statement

Retail investors face a large catalog of financial assets and must combine risk, return, sector, valuation and personal preferences. Generic “top stock” lists are not personalized. This project studies a hybrid recommender that ranks assets for an investor profile.


## 2. Research question and hypotheses

**Research question:** Can a hybrid two-tower neural recommender outperform global-popularity and content-based baselines on temporally held-out user–asset interactions?

- **H1:** Personalization improves Hit Rate@10 over global popularity.
- **H2:** Combining identity embeddings with user and asset features improves catalog coverage.
- **H3:** A temporal user-level split enables leakage-aware evaluation by
  ensuring that model training uses only interactions occurring before
  validation and test interactions.

## 3. Scope and practical constraints

The starter version uses deterministic synthetic investors and fictional assets. This avoids licensing and API instability and makes every result reproducible. The synthetic interaction mechanism deliberately links risk profiles and sector preferences to asset characteristics. Therefore, good performance proves that the pipeline can recover the simulated preference structure; it does **not** prove real-world investment value.

## 4. Previous research

1. Koren, Bell and Volinsky (2009), *Matrix Factorization Techniques for Recommender Systems*. Establishes latent-factor baselines.
2. He et al. (2017), *Neural Collaborative Filtering*. Replaces the fixed interaction function of matrix factorization with neural networks.
3. Covington, Adams and Sargin (2016), *Deep Neural Networks for YouTube Recommendations*. Describes candidate generation with separate user and item representations.
4. Rendle et al. (2012), *BPR: Bayesian Personalized Ranking from Implicit Feedback*. Motivates ranking from implicit interactions.
5. TensorFlow Recommenders documentation and examples. Provides retrieval/ranking implementation patterns for future extension.

The current project compares a two-tower neural model against popularity, sector-aware popularity and content similarity.

## 5. Methodology overview

1. Generate or load user, asset and interaction tables.
2. Validate identifiers, types and missing values.
3. Explore user and asset distributions.
4. Split each user's interactions chronologically.
5. Train baselines and a two-tower model.
6. Evaluate Top-K ranking quality and catalog coverage.
7. Interpret limitations and propose improvements.
